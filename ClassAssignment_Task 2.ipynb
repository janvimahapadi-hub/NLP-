```python
import pandas as pd
import string
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
nltk.download("stopwords")
nltk.download("wordnet")
```

```python
documents = [
    {
        "title": "Data Science Intern",
        "description": "Looking for Python, machine learning, data analysis, pandas, NumPy, and SQL skills."
    },
    {
        "title": "Frontend Developer",
        "description": "Experience with HTML, CSS, JavaScript, React, and responsive web design."
    },
    {
        "title": "Cybersecurity Analyst",
        "description": "Knowledge of network security, vulnerability assessment, penetration testing, and risk analysis."
    },
    {
        "title": "Machine Learning Engineer",
        "description": "Experience with Python, TensorFlow, deep learning, artificial intelligence, and model building."
    },
    {
        "title": "Data Analyst",
        "description": "Strong SQL, Excel, Power BI, Tableau, data visualization, and reporting skills."
    },
    {
        "title": "Backend Developer",
        "description": "Knowledge of Python, APIs, databases, Django, Flask, and server-side development."
    },
    {
        "title": "AI Research Intern",
        "description": "Work on natural language processing, machine learning, artificial intelligence, and research projects."
    },
    {
        "title": "Cloud Engineer",
        "description": "Experience with AWS, cloud infrastructure, Docker, Kubernetes, and deployment."
    },
    {
        "title": "Business Analyst",
        "description": "Analyze business data, create reports, gather requirements, and improve decision making."
    },
    {
        "title": "DevOps Engineer",
        "description": "Experience with CI CD pipelines, Docker, Linux, automation, monitoring, and cloud systems."
    }
]

df = pd.DataFrame(documents)
```

```python
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    # lowercase
    text = text.lower()

    # remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # split text into words
    words = text.split()

    # remove stopwords and lemmatize
    cleaned_words = []
    for word in words:
        if word not in stop_words:
            word = lemmatizer.lemmatize(word)
            cleaned_words.append(word)

    return " ".join(cleaned_words)


df["cleaned_description"] = df["description"].apply(preprocess_text)
```

```python
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df["cleaned_description"])
```

```python
query = "python machine learning data analysis"
cleaned_query = preprocess_text(query)
```

```python
query_vector = vectorizer.transform([cleaned_query])
```

```python
similarity_scores = cosine_similarity(query_vector, tfidf_matrix).flatten()
```

```python
df["similarity_score"] = similarity_scores
```

```python
top_results = df.sort_values(by="similarity_score", ascending=False).head(3)
```

```python
print("\nTop 3 Recommended Jobs:\n")
print("User Query:", query)
print("Processed Query:", cleaned_query)

for index, row in top_results.iterrows():
    print("Title:", row["title"])
    print("Description:", row["description"])
    print("Similarity Score:", round(row["similarity_score"], 3))
    print("-" * 60)
```
